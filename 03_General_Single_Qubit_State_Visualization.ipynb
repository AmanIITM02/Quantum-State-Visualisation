```python
!pip install qutip
!pip install numpy

import qutip as qt
import numpy as np

#Defining the Probability Amplitudes

alpha = 0.8
beta = 0.6

#Verifying the Normalization
normalization = abs(alpha)**2 + abs(beta)**2
print("Normalization =", normalization)
if not np.isclose(normalization, 1.0):
    raise ValueError("The Quantum State is not Normalized")

#Creating Computational Basis States
zero = qt.basis(2, 0)    #|0>
one = qt.basis(2, 1)     #|1>

#Construction part of the General Quantum State
psi = alpha * zero + beta * one
print("\nQuantum State:")
print(psi)

#Calculation of Measurement Possibilities
prob_zero = abs(alpha)**2
prob_one = abs(beta)**2

#Locating the Bloch Vector Coordinates
x = 2 * np.real(np.conj(alpha) * beta)
y = 2 * np.imag(np.conj(alpha) * beta)
z = abs(alpha)**2 - abs(beta)**2

print("\nBloch Vector:")
print("x = ", x)
print("y = ", y)
print("z = ", z)

#Final Output of the Bloch Sphere
bloch = qt.Bloch()
bloch.add_states(psi)
bloch.show()
```
