```python
#Quantum State Visualization : Superposition of the Qubit

#Installing the Qutip
!pip install qutip
import qutip as qt
import numpy as np

#Creating Computational basis states
zero = qt.basis(2, 0)   #|0>
one = qt.basis(2, 1)    #|1>

#Creating the equal superposition state
#|superposition> = (|0> + |1>)/√2

superposition = (zero + one)/np.sqrt(2)

#Bloch sphere creation
bloch = qt.Bloch()

#Embedding the Superposition State to the Bloch Sphere
bloch.add_states(superposition)
bloch.show()
```
